```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="darkgrid")
plt.figure(figsize=(12, 6))
sns.barplot(x='Surgical_Type', y='Recovery_Time_Min', data=df, errorbar=None, palette='rocket')
plt.title('Average Recovery Time by Surgical Type', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Surgical Type', fontsize=11)
plt.ylabel('Average Recovery Time (Minutes)', fontsize=11)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
plt.figure(figsize=(9, 5))
sns.scatterplot(x='Dosage_mg', y='Recovery_Time_Min', hue='Complications', data=df, palette='viridis', s=80, alpha=0.8)
plt.title('Anesthesia Dosage vs. Recovery Time by Complications', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Anesthesia Dosage (mg)', fontsize=11)
plt.ylabel('Recovery Time (Minutes)', fontsize=11)
plt.legend(title='Complications Happened?')
plt.tight_layout()
plt.show()
```
